# PIANO — Composer Parameters

Key parameters and their defaults. Full list in `piano/utils/composer.py`.

## Data & memory

| Parameter | Default | Description |
|---|---|---|
| `adata` | — | AnnData or list of AnnDatas |
| `memory_mode` | `'GPU'` | `'GPU'`, `'SparseGPU'`, `'CPU'`, `'SparseCPU'`, `'backed'` |
| `compile_model` | `True` | Use `torch.compile` |
| `categorical_covariate_keys` | `None` | List of obs columns to treat as categorical |
| `continuous_covariate_keys` | `None` | List of obs columns to treat as continuous |
| `sparse_continuous_covariate_keys` | `None` | List of `(category, value)` column pairs, e.g. drug + dose |

## Gene selection

| Parameter | Default | Description |
|---|---|---|
| `n_top_genes` | `4096` | HVGs to select. `-1` = use all genes |
| `flavor` | `'seurat_v3'` | HVG method (only Seurat v3 supported across multiple AnnDatas) |
| `hvg_batch_key` | `None` | Batch key for HVG selection |
| `geneset_path` | `None` | Path to text file with one gene per line |

## Model architecture

| Parameter | Default | Description |
|---|---|---|
| `n_hidden` | `256` | Hidden layer size |
| `n_layers` | `3` | Encoder/decoder depth |
| `latent_size` | `32` | Latent space dimensions |
| `distribution` | `'nb'` | `'nb'` (negative binomial) or `'zinb'` |
| `parameterization` | `'ksi-psi'` | Negative binomial parameterization: `'ksi-psi'` or `'mu-theta'` |
| `dropout_rate` | `0.1` | Dropout probability |

## Training

| Parameter | Default | Description |
|---|---|---|
| `stratify_column` | `None` | obs column to balance training on |
| `samples_per_class` | `1000` | Max cells per class per epoch (only used with `stratify_column`; `None` = all cells) |
| `max_epochs` | `200` | Training epochs |
| `batch_size` | `128` | Mini-batch size |
| `lr` | `2e-4` | Learning rate |
| `adversarial` | `True` | Enable adversarial batch correction |
| `min_kld_weight` | `0.00` | KL weight start (annealing) |
| `max_kld_weight` | `0.25` | KL weight ceiling |
| `n_annealing_epochs` | `None` | Epochs to ramp KL/adversarial weights (`None` = `max_epochs`) |
| `min_adv_weight` | `1.00` | Adversarial weight start |
| `max_adv_weight` | `1.00` | Adversarial weight ceiling |
| `early_stopping` | `True` | Stop training when loss plateaus |
| `patience` | `5` | Epochs without improvement before stopping |

## Reproducibility & output

| Parameter | Default | Description |
|---|---|---|
| `deterministic` | `True` | Fix random ops for reproducibility. On GPU, set `os.environ["CUBLAS_WORKSPACE_CONFIG"] = ":4096:8"` before importing piano |
| `random_seed` | `0` | Global seed |
| `run_name` | `'piano_integration'` | Prefix for saved files |
| `outdir` | `'./results/'` | Output directory |

## Full parameter reference

```python
from piano import Composer

pianist = Composer(
    # Data
    adata=adata,
    memory_mode='GPU',             # 'GPU' | 'SparseGPU' | 'CPU' | 'SparseCPU' | 'backed'
    compile_model=True,
    categorical_covariate_keys=None,
    continuous_covariate_keys=None,
    sparse_continuous_covariate_keys=None,
    unlabeled='Unknown',

    # Gene selection
    flavor='seurat_v3',
    n_top_genes=4096,
    hvg_batch_key=None,
    geneset_path=None,

    # Architecture
    n_hidden=256,
    n_layers=3,
    latent_size=32,
    distribution='nb',             # 'nb' | 'zinb'
    parameterization='ksi-psi',    # 'ksi-psi' | 'mu-theta'
    dropout_rate=0.1,
    batchnorm_eps=1e-5,
    batchnorm_momentum=1e-1,
    epsilon=1e-5,

    # Training mode
    adversarial=True,

    # Training
    stratify_column=None,
    samples_per_class=1000,
    max_epochs=200,
    batch_size=128,
    min_kld_weight=0.00,
    max_kld_weight=0.25,
    min_adv_weight=1.00,
    max_adv_weight=1.00,
    n_annealing_epochs=None,
    lr=2e-4,
    weight_decay=0.00,
    shuffle=True,
    drop_last=True,
    num_workers=0,

    # Early stopping
    early_stopping=True,
    min_delta=1.00,
    patience=5,

    # Checkpoints
    save_initial_weights=False,
    checkpoint_every_n_epochs=None,

    # Reproducibility
    deterministic=True,
    random_seed=0,

    # Output
    run_name='piano_integration',
    outdir='./results/',
)
```
