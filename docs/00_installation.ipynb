# Installation

This notebook covers how to install `piano-integration` and verify your environment.

- **Recommended:** `uv` (fast, reproducible)
- **Alternative:** `conda`
- **GPU acceleration:** RAPIDS optional but strongly recommended for large datasets (needs an NVIDIA Volta or newer GPU, compute capability ≥ 7.0)

> Python ≥ 3.11 is required.

## 1. Install uv

```python
# Run in your terminal (not inside a notebook)
curl -LsSf https://astral.sh/uv/install.sh | sh

# Optionally set a cache directory
echo 'cache-dir = "/path/to/cache/uv_cache"' >> ~/.config/uv/uv.toml
```

## 2. Create a virtual environment

```python
uv venv --python 3.11.14
source .venv/bin/activate
```

## 3. Install piano-integration

Choose the variant that matches your hardware:

```python
# Standard (CPU)
uv pip install piano-integration

# With RAPIDS (GPU — recommended for large datasets)
uv pip install piano-integration[rapids]

# All optional libraries
uv pip install piano-integration[all]

# If you hit dependency conflicts, add:
# --index-strategy unsafe-best-match
```

## 4. conda alternative (slower)

```python
conda create -n piano python=3.11.14 -y
conda activate piano
pip install piano-integration[rapids]
```

## 5. ARM / Triton compilation

Only needed on ARM architectures (e.g. Apple Silicon, ARM HPC nodes) for `torch.compile` support.

```python
git clone https://github.com/triton-lang/triton.git
cd triton
uv pip install -r python/requirements.txt
uv pip install -e .
```

## 6. Verify installation

```python
import piano
import torch

print(f"piano-integration version: {piano.__version__}")
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
```
